```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
penguin = pd.read_csv(url)

fitur = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
penguin = penguin.dropna(subset=fitur)

X = penguin[fitur]
y = penguin["species"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=42)

model = GaussianNB()
model.fit(X_train, y_train)
```

#Latihan 1

```python
m1 = GaussianNB()
m1.fit(X_train[["flipper_length_mm"]], y_train)
y_pred1 = m1.predict(X_test[["flipper_length_mm"]])
print(f"Akurasi 1 fitur: {accuracy_score(y_test, y_pred1):.3f}")
print(classification_report(y_test, y_pred1, digits=3))
```

Akurasi turun karena pada histogram panjang sirip, Adelie dan Chinstrap hampir menumpuk. Hanya Gentoo yang terpisah jelas. Panjang paruh adalah fitur yang membedakan Adelie dan Chinstrap, jadi model satu fitur sering salah di dua spesies itu.

#Latihan 2

```python
baru = pd.DataFrame([{
    "bill_length_mm": 47.5,
    "bill_depth_mm": 15.0,
    "flipper_length_mm": 217.0,
    "body_mass_g": 5000.0,
}])
proba = model.predict_proba(baru)[0]
for s, p in zip(model.classes_, proba):
    print(f"{s:10s}: {p:.3f}")
print("Prediksi:", model.predict(baru)[0])
```

Hasilnya Gentoo, sesuai histogram: sirip panjang (217 mm), badan berat (5000 g), dan paruh tipis (15 mm) hanya cocok dengan sebaran Gentoo.
