```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report

url = ("https://raw.githubusercontent.com/ahmad-fathan/Fundamentals-of-Data-Science/"
       "main/Classification/diabetes_prediction_dataset.csv.gz")
diabetes = pd.read_csv(url)

fitur = ["age", "hypertension", "heart_disease", "bmi", "HbA1c_level", "blood_glucose_level"]
X = diabetes[fitur]
y = diabetes["diabetes"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)

model = GaussianNB()
model.fit(X_train, y_train)
```

#Latihan 1

```python
m_hba = GaussianNB()
m_hba.fit(X_train[["HbA1c_level"]], y_train)
y_pred_hba = m_hba.predict(X_test[["HbA1c_level"]])
print(f"Akurasi: {accuracy_score(y_test, y_pred_hba):.3f}")
print(classification_report(y_test, y_pred_hba, digits=3,
      target_names=["tidak diabetes", "diabetes"]))
```

Akurasi naik karena model satu fitur lebih jarang salah menandai pasien sehat sebagai diabetes, dan 91,5% data memang sehat. Tetapi recall diabetes turun jauh (0,66 jadi sekitar 0,23) karena banyak pasien diabetes terlewat. Pada data tidak seimbang, akurasi tidak cukup. Untuk skrining, recall lebih penting.

#Latihan 2

```python
m_glu = GaussianNB()
m_glu.fit(X_train[["blood_glucose_level"]], y_train)
y_pred_glu = m_glu.predict(X_test[["blood_glucose_level"]])
print(f"Akurasi: {accuracy_score(y_test, y_pred_glu):.3f}")
print(classification_report(y_test, y_pred_glu, digits=3,
      target_names=["tidak diabetes", "diabetes"]))
```

**Hasil Latihan 2 (blood_glucose_level):** akurasi 0,948, precision diabetes 1,000, dan recall diabetes 0,389.

**Perbandingan dengan Latihan 1 (HbA1c_level):** HbA1c menghasilkan akurasi 0,934, precision diabetes 1,000, dan recall diabetes 0,228. Fitur blood_glucose_level lebih berguna jika dipakai sendirian karena akurasinya lebih tinggi dan recall diabetesnya lebih besar (0,389 dibanding 0,228), artinya lebih banyak pasien diabetes yang berhasil ditemukan.

**Kesesuaian dengan histogram E3.3:** Pada histogram, puncak pasien diabetes bergeser ke kanan untuk kedua fitur, dan sebarannya cukup terpisah dari pasien tidak diabetes. Hasil angka sesuai dengan pengamatan itu, karena kedua fitur memberi akurasi di atas 0,93 dan precision diabetes 1,000. Namun kedua model masih melewatkan banyak pasien diabetes (recall di bawah 0,4), karena model satu fitur tidak cukup untuk menangkap semua kasus.

**Kesimpulan:** Karena data tidak seimbang (91,5% pasien tidak diabetes), akurasi saja tidak cukup. Untuk skrining, recall kelas diabetes lebih penting, dan blood_glucose_level lebih baik dibanding HbA1c_level dalam hal itu. Meski begitu, recall 0,389 masih rendah, sehingga model dengan enam fitur (recall sekitar 0,66) tetap lebih baik untuk menemukan pasien diabetes.
